```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


class SalesDataAnalyzer:

    def __init__(self):
        self.data = None

    # Load Dataset
    def load_data(self):
        try:
            self.data = pd.read_csv("sales_data.csv")
            print("Dataset loaded successfully!")
            print(self.data.head())
        except FileNotFoundError:
            print("CSV file not found!")

    # Explore Data
    def explore_data(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        print("\n1. First 5 rows")
        print("2. Last 5 rows")
        print("3. Column names")
        print("4. Data types")

        choice = input("Enter your choice: ")

        if choice == "1":
            print(self.data.head())
        elif choice == "2":
            print(self.data.tail())
        elif choice == "3":
            print(self.data.columns)
        elif choice == "4":
            print(self.data.dtypes)
        else:
            print("Invalid choice!")

    # Clean Data
    def clean_data(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        self.data = self.data.drop_duplicates()
        self.data = self.data.fillna(0)
        print("Data cleaned successfully!")

    # DataFrame Operations
    def dataframe_operations(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        print("\n1. Search product")
        print("2. Sort sales")
        print("3. Filter region")
        print("4. Total sales")

        choice = input("Enter your choice: ")

        if choice == "1":
            product = input("Enter product name: ")
            print(self.data[self.data["Product"].str.contains(
                product, case=False, na=False
            )])

        elif choice == "2":
            print(self.data.sort_values("Sales", ascending=False))

        elif choice == "3":
            region = input("Enter region: ")
            print(self.data[self.data["Region"].str.contains(
                region, case=False, na=False
            )])

        elif choice == "4":
            print("Total Sales:", self.data["Sales"].sum())

        else:
            print("Invalid choice!")

    # Statistics
    def statistics(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        print("\nSales Statistics")
        print(self.data["Sales"].describe())
        print("Mean:", self.data["Sales"].mean())
        print("Variance:", self.data["Sales"].var())
        print("Standard Deviation:", self.data["Sales"].std())

    # Visualization
    def visualization(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        print("\n1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Heatmap")

        choice = input("Enter your choice: ")

        if choice == "1":
            self.data.groupby("Product")["Sales"].sum().plot(kind="bar")
            plt.title("Sales by Product")

        elif choice == "2":
            self.data.groupby("Year")["Sales"].sum().plot(
                kind="line", marker="o"
            )
            plt.title("Sales by Year")

        elif choice == "3":
            plt.scatter(self.data["Sales"], self.data["Profit"])
            plt.xlabel("Sales")
            plt.ylabel("Profit")
            plt.title("Sales vs Profit")

        elif choice == "4":
            self.data.groupby("Region")["Sales"].sum().plot(
                kind="pie", autopct="%1.1f%%"
            )
            plt.title("Sales by Region")
            plt.ylabel("")

        elif choice == "5":
            plt.hist(self.data["Sales"], bins=5, edgecolor="black")
            plt.title("Sales Histogram")

        elif choice == "6":
            sns.heatmap(
                self.data.select_dtypes(include=np.number).corr(),
                annot=True
            )
            plt.title("Sales Heatmap")

        else:
            print("Invalid choice!")
            return

        plt.tight_layout()
        plt.show()

    # Save Visualization
    def save_visualization(self):
        if self.data is None:
            print("Please load the dataset first!")
            return

        self.data.groupby("Product")["Sales"].sum().plot(kind="bar")
        plt.title("Sales by Product")
        plt.tight_layout()
        plt.savefig("sales_plot.png")
        plt.show()
        print("Chart saved as sales_plot.png")

    # Main Menu
    def run(self):
        while True:
            print("\n===== SALES DATA ANALYZER =====")
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. DataFrame Operations")
            print("4. Clean Data")
            print("5. Generate Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.load_data()
            elif choice == "2":
                self.explore_data()
            elif choice == "3":
                self.dataframe_operations()
            elif choice == "4":
                self.clean_data()
            elif choice == "5":
                self.statistics()
            elif choice == "6":
                self.visualization()
            elif choice == "7":
                self.save_visualization()
            elif choice == "8":
                print("Thank you! Program closed.")
                break
            else:
                print("Please enter a choice from 1 to 8.")


# Start Program
project = SalesDataAnalyzer()
project.run()
```
